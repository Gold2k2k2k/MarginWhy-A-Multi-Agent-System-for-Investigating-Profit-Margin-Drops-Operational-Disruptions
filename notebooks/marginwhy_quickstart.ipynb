### 4. Interactive Programmatic Inspection (Python API)

Directly inspect the H-MEM memory state and Markovian carryover.

```python
from app.agents.orchestrator import MarginWhyOrchestrator
import json

orchestrator = MarginWhyOrchestrator(sample_mode=True)
result = orchestrator.run_investigation(
    query="Why did net profit margin for Electronics drop 18.5% in Q3?",
    scenario_file="scenario_1_electronics_q3.json"
)

print("Turns Executed:", result["turns_executed"])
print("Approved Claim:", result["approved_claims"][0]["hypothesis_id"])
print("Confidence Score:", result["approved_claims"][0]["confidence_score"])
print("\n--- Markovian Bounded Carryover ---")
print(json.dumps(result["hmem_state"], indent=2))

print("\n--- Synthesized Executive Report ---")
print(result["report_markdown"][:500] + "...")
```

### 5. Inspect Audit Trail & NVIDIA Nemotron Model Tagging

```python
with open("logs/trace.jsonl", "r", encoding="utf-8") as f:
    events = [json.loads(line) for line in f if line.strip()]

nemotron_events = [e for e in events if "nemotron" in str(e.get("model", "")).lower()]
blocked_events = [e for e in events if e.get("event") == "CLAIM_BLOCKED"]
approved_events = [e for e in events if e.get("event") == "APPROVED_CLAIM"]

print(f"Total trace events recorded: {len(events)}")
print(f"Events tagged with NVIDIA Nemotron model: {len(nemotron_events)}")
print(f"Hypothesis rejection events (CLAIM_BLOCKED): {len(blocked_events)}")
print(f"Approved claims (APPROVED_CLAIM): {len(approved_events)}")

print("\nSample CLAIM_BLOCKED event:")
print(json.dumps(blocked_events[-1], indent=2))
```
